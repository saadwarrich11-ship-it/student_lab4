## Part 12: Performance — loops versus NumPy

All timings are measured at runtime; rerun this notebook on your own device to get your own numbers.

```python
import time, timeit, tracemalloc, csv
from pathlib import Path
import sys
from pathlib import Path
sys.path.insert(0, str(Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()))
from src.marks_utils import standardise_loop, standardise_np
def loop_double(values):
    result=[]
    for x in values: result.append(x*2)
    return result
def numpy_double(values): return values*2
N=1_000_000
py_list=list(range(N)); np_arr=np.arange(N)
for label,func,values in [('loop',loop_double,py_list),('numpy',numpy_double,np_arr)]:
    start=time.perf_counter(); func(values)
    print(label,round(time.perf_counter()-start,5),"seconds")
n=20
t_loop=timeit.timeit(lambda:loop_double(py_list),number=n)/n
t_np=timeit.timeit(lambda:numpy_double(np_arr),number=n)/n
print("timeit",t_loop,t_np,t_loop/t_np)
for label,make in [('list',lambda:[x*2 for x in range(N)]),('array',lambda:np.arange(N)*2)]:
    tracemalloc.start(); obj=make(); current,peak=tracemalloc.get_traced_memory(); tracemalloc.stop()
    print(label,"traced peak MB",round(peak/1e6,2))
    del obj
```

```python
rows=[]
for size in [1_000,100_000,1_000_000]:
    li=list(range(size)); ar=np.arange(size)
    count=100 if size<1_000_000 else 20
    tl=timeit.timeit(lambda:loop_double(li),number=count)/count
    tn=timeit.timeit(lambda:numpy_double(ar),number=count)/count
    rows.append((size,tl,tn,tl/tn))
print("N, loop_s, numpy_s, ratio")
for row in rows: print(*row,sep=", ")
print("Best of 5 (one million):")
print("loop",min(timeit.repeat(lambda:loop_double(py_list),number=5,repeat=5))/5)
print("numpy",min(timeit.repeat(lambda:numpy_double(np_arr),number=5,repeat=5))/5)
print("Conclusion based on these measured results: speed ratios =",[round(r[3],2) for r in rows])
print("Best-of-several reduces impact of interruptions and transient slow runs.")
```

```python
data=np.random.default_rng(0).random((20_000,10))
loop_result=standardise_loop(data)
np_result=standardise_np(data)
print("equivalent",np.allclose(loop_result,np_result))
for label,fn in [('loop',standardise_loop),('numpy',standardise_np)]:
    t=time.perf_counter(); fn(data); print(label,"seconds",time.perf_counter()-t)
    tracemalloc.start(); result=fn(data); current,peak=tracemalloc.get_traced_memory(); tracemalloc.stop()
    print(label,"traced peak MB",peak/1e6)
out=Path('../data/processed') if Path('../data/processed').exists() else Path('data/processed')
out.mkdir(parents=True,exist_ok=True)
with (out/'timing_results.csv').open('w',newline='') as f:
    writer=csv.writer(f); writer.writerow(['N','loop_s','numpy_s','ratio']); writer.writerows(rows)
np.save(out/'standardised.npy',np_result)
print("Saved",out)
```
